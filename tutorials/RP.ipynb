```python
# Random Projection
# Note: There are two types of random projection: Gaussian and Sparse. The Gaussian random projection is more accurate but slower, while the sparse random projection is faster but less accurate. We will implement both using the sklearn library.
```

```python
# import statements
import numpy as np
from sklearn.random_projection import GaussianRandomProjection, SparseRandomProjection
```

```python
# generating random dataset (100 samples, 5000 features)
rng = np.random.RandomState(42) # random seed for the data
X = rng.rand(100, 5000) # 100 samples, 5000 features
```

```python
# Gaussian Random Projection
# projects data using a matrix with normally distributed elements
transformer_gaussian = GaussianRandomProjection(n_components=500, random_state=42)
X_gaussian = transformer_gaussian.fit_transform(X)

# Sparse Random Projection
# projects data using a matrix with sparse elements
# this is more memeory-efficient as it uses mostly zero values with occasional +1 or -1
transformer_sparse = SparseRandomProjection(n_components=500, random_state=42)
X_sparse = transformer_sparse.fit_transform(X)

print("Original shape:", X.shape) # (100, 5000)
print("Gaussian Random Projection shape:", X_gaussian.shape) # (100, 500)
print("Sparse Random Projection shape:", X_sparse.shape) # (100, 500)
```
